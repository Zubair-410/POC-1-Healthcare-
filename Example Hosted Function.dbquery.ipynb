```sql
-- This is an example of a hosted function that will allow your agent to query a table and retrieve useful insights.
-- Select the catalog and schema you want to save the function to and run it.
-- Then go back to Playground an add the function by finding it with "Add existing tool"
CREATE OR REPLACE FUNCTION avg_fare_by_distance(min_miles DOUBLE, max_miles DOUBLE)
RETURNS DOUBLE
LANGUAGE SQL
COMMENT 'Returns the average taxi fare amount for trips with a distance between min_miles and max_miles.'
RETURN (
  SELECT AVG(fare_amount)
  FROM samples.nyctaxi.trips
  WHERE trip_distance BETWEEN min_miles AND max_miles
);
```
