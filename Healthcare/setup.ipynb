```python
%sql
CREATE SCHEMA IF NOT EXISTS healthcare.bronze;

CREATE SCHEMA IF NOT EXISTS healthcare.silver;

CREATE SCHEMA IF NOT EXISTS healthcare.gold;

CREATE SCHEMA IF NOT EXISTS healthcare.control;
```

```python
%sql
CREATE TABLE IF NOT EXISTS healthcare.control.pipeline_batches (

    batch_id STRING,
    pipeline_name STRING,
    start_time TIMESTAMP,
    end_time TIMESTAMP,
    status STRING,
    source_files INT,
    processed_records BIGINT

);
```

```python
%sql
CREATE TABLE IF NOT EXISTS healthcare.bronze.bronze_patients
(
    patient_id STRING,
    name STRING,
    age INT,
    arrival_date DATE,
    departure_date DATE,
    service STRING,
    satisfaction INT,

    file_name STRING,
    file_path STRING,
    file_size BIGINT,
    file_modification_time TIMESTAMP,

    source_format STRING,

    batch_id STRING,
    processed_timestamp TIMESTAMP,
    pipeline_stage STRING
);
```

```python
%sql
CREATE TABLE IF NOT EXISTS healthcare.bronze.bronze_staff
(
    staff_id STRING,
    staff_name STRING,
    role STRING,
    service STRING,

    file_name STRING,
    file_path STRING,
    file_size BIGINT,
    file_modification_time TIMESTAMP,

    source_format STRING,

    batch_id STRING,
    processed_timestamp TIMESTAMP,
    pipeline_stage STRING
);
```

```python
%sql
CREATE TABLE IF NOT EXISTS healthcare.bronze.bronze_staff_schedule
(
    week INT,
    staff_id STRING,
    staff_name STRING,
    role STRING,
    service STRING,
    present INT,

    file_name STRING,
    file_path STRING,
    file_size BIGINT,
    file_modification_time TIMESTAMP,

    source_format STRING,

    batch_id STRING,
    processed_timestamp TIMESTAMP,
    pipeline_stage STRING
);
```

```python
%sql
CREATE TABLE IF NOT EXISTS healthcare.bronze.bronze_services_weekly
(
    week INT,
    month INT,
    service STRING,

    available_beds INT,
    patients_request INT,
    patients_admitted INT,
    patients_refused INT,

    patient_satisfaction INT,
    staff_morale INT,

    event STRING,

    file_name STRING,
    file_path STRING,
    file_size BIGINT,
    file_modification_time TIMESTAMP,

    source_format STRING,

    batch_id STRING,
    processed_timestamp TIMESTAMP,
    pipeline_stage STRING
);
```
